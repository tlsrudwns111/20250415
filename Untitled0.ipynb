```python
import math


PI = math.pi
g = -9.8

V0 = float(input("초기속도 입력: "))
theta_degree = float(input("각도 입력: "))


rad = math.radians(theta_degree)


V0x = V0 * math.cos(rad)
V0y = V0 * math.sin(rad)

print(f"x방향의 초기속도 : {V0x:.2f}")
print(f"y방향의 초기속도 : {V0y:.2f}")


h = (V0y ** 2) / (2 * -g)
print(f"최고점의 높이 : {h:.2f}")


t = V0y / -g
print(f"최고점에 도달하는 시간은 : {t:.2f}")


s = V0x * t
print(f"수평으로 이동한 최고 거리는 : {s:.2f}")


total_time = 2 * t
Vy_final = V0y + g * total_time
Vx_final = V0x

V_final = math.sqrt(Vx_final**2 + Vy_final**2)
print(f"지면에 닿을때의 Vx = {Vx_final:.2f}, Vy = {Vy_final:.2f}, V_final = {V_final:.2f}")
```
