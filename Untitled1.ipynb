```python
import math

# 입력 받기
v0 = float(input("초기 속도: "))
angle = float(input("발사 각도: "))
y0 = float(input("초기 높이: "))

pi = math.pi
g = 9.81

# 각도를 라디안으로 변환
rad = angle * pi / 180.0

# 속도의 x, y 성분 계산
v0x = v0 * math.cos(rad)
v0y = v0 * math.sin(rad)

# 2차 방정식의 계수
a = -0.5 * g
b = v0y
c = y0

# 판별식
D = b**2 - 4*a*c

if D < 0:
    print("오류: 실근이 존재하지 않아요 ㅠㅠ")
else:
    t1 = (-b + math.sqrt(D)) / (2 * a)
    t2 = (-b - math.sqrt(D)) / (2 * a)
    t3 = max(t1, t2)  # 양의 시간 선택

    print(f"(a) 지면에 도달할 때까지의 시간: {t3:.2f} 초")

    # 수평 거리
    d = v0x * t3
    print(f"(b) 수평 거리: {d:.2f} m")

    # 속도 성분
    vx = v0x
    vy = v0y - g * t3
    print(f"(c) 속도의 수평 성분: {vx:.2f} m/s")
    print(f"(c) 속도의 수직 성분: {vy:.2f} m/s")

    # 속도 크기
    s = math.sqrt(vx**2 + vy**2)
    print(f"(d) 속도의 크기: {s:.2f} m/s")

    # 수평과의 각도
    ang = math.atan2(vy, vx) * 180.0 / pi
    print(f"(e) 수평 방향과 이루는 각도: {ang:.2f} 도")
```
